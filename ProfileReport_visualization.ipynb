**What is the distribution of passenger ages on the Titanic?**

```python
data = {
    'Gender': np.random.choice(['Male', 'Female'], size=100),
    'Age_Group': np.random.choice(['18-25', '26-35', '36-45', '46-55'], size=100),
    'Exercise_Frequency': np.random.choice(['Rarely', 'Sometimes', 'Regularly'], size=100),
    'Commute_Method': np.random.choice(['Car', 'Public Transport', 'Walking/Cycling'], size=100)
}

df = pd.DataFrame(data)
df.head()
```

**Based on the given DataFrame, which plot will be visualized to show the relationship between Gender and Exercise_Frequency?**

```python
plt.figure(figsize=(8, 5))
# Fill the missing line to plot the chart
sns.countplot(data = df, x= 'Exercise_Frequency',hue = "Gender")
plt.title('Gender vs. Exercise Frequency')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Exercise_Frequency', fontsize='small')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Create a plot to show the progress of likes and views over time. Then perform given stylings**

```python
dates = pd.date_range(start='2022-01-01', end='2022-01-31')
video_views = np.random.randint(1000, 2000, size=len(dates))
video_likes = np.random.randint(100, 1000, size=len(dates))


daily_progress = pd.DataFrame({'Date': dates, 'Views': video_views, 'Likes': video_likes})
daily_progress.head()
```

```python
# Plotting
plt.figure(figsize=(10, 6))
#Write the code
plt.plot(daily_progress['Date'], daily_progress['Views'], label='Views', marker='*', linestyle='-.',markersize=8)
plt.plot(daily_progress['Date'], daily_progress['Likes'], label='Likes', marker='*', linestyle='-.',markersize=8)

plt.title('Daily Progress of YouTube Channels - Views Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Views')
plt.legend()
#Write the code rotate the xticks
plt.show()
```

**Plot relationship between two variables. Then do given stylings )**
    
    1. Add legends, customize legends' colors
    2. Use a marker other than 'o', '*', and 'h',
    3. Change the edgecolor of marker
    4. Change the markersize
    5. Change the color of labels
    6. Change the color of ticks

```python
x = np.random.rand(100)
y = np.random.rand(100)
colors = np.random.rand(100)
sizes = np.random.randint(10, 100, 100)
```

```python
plt.figure(figsize=(8, 6))
#Create a plot
scatter = plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, marker='p', edgecolor='cyan')
# Adding legend with customized colors
plt.legend(handles=scatter.legend_elements()[0], labels=['Random Points'], loc='best')
# Change the legend label color
plt.gca().get_legend().legendHandles[0].set_color('blue')
plt.title('Scatter Plot of Random Data')
plt.title('Scatter Plot of Random Data')

# Customizing labels, x-y ticks and their colors
plt.xlabel('X_Axis', color='black')
plt.ylabel('Y_Axis', color='green')
plt.xticks(color='blue')
plt.yticks(color='red')
plt.show()

```
